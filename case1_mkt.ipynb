```sql
-- ============= CATEGORY SECTIONS =============

DROP TABLE IF EXISTS #temp_category_section;

SELECT r.rental_id as rental_id,
    r.customer_id as customer_id,
    fc.category_id as category_id,
    c.name as category_name,
    f.film_id as film_id,
    f.title as film_title,
    r.rental_date as rental_date
INTO #temp_category_section
FROM danny_mktcase.rental r 
LEFT JOIN danny_mktcase.inventory i ON r.inventory_id = i.inventory_id
LEFT JOIN danny_mktcase.film_category fc ON i.film_id = fc.film_id
LEFT JOIN danny_mktcase.category c ON c.category_id = fc.category_id
LEFT JOIN danny_mktcase.film f ON fc.film_id = f.film_id

-- Select top 5 from #temp_category_section
SELECT TOP 5 * FROM #temp_category_section
```

```sql
-- CREATE TABLE: TOP 2 CATEGORIES
DROP TABLE IF EXISTS #temp_top2_categories; 

---- Calculate count rental for each customer by categories
WITH cte_rentals_by_customer AS (
    SELECT customer_id as customer_id,
        category_id as category_id,
        category_name as category_name,
        COUNT(rental_id) as count_rentals,
        MAX(rental_date) as lastest_rental_date,
        PERCENT_RANK() OVER(PARTITION BY category_id ORDER BY COUNT(rental_id) DESC) as percent_rank
        
    FROM #temp_category_section 
    GROUP BY customer_id, category_id, category_name
),
---- Calculate average rental count by each category
cte_avg_rental_by_category AS (
    SELECT c.category_id,
        AVG(count_rentals) as avg_rentals_category

    FROM cte_rentals_by_customer c
    GROUP BY c.category_id

),
---- Check max rentals count by each category
cte_max_rentals_by_category AS (
    SELECT c.category_id,
        MAX(c.count_rentals) as max_count_rentals
    FROM cte_rentals_by_customer c
    GROUP BY c.category_id
),
cte_pct_rental AS (
    SELECT c.customer_id,
        SUM(c.count_rentals) as total_rental_count_by_customer
    FROM cte_rentals_by_customer c
    GROUP BY c.customer_id
)
---- Top 2 categories rentals count
SELECT c.customer_id,
    c.category_id,
    c.category_name,
    c.count_rentals,
    c.category_ranking,
    (c.count_rentals - c.avg_rentals_category) as diff_rental_count_category,
    c.total_rental_count_by_customer,
    ROUND(100 * c.count_rentals / c.total_rental_count_by_customer, 0) as rental_share,
    CASE 
        WHEN ROUND(c.percent_rank * 100, 0) = 0 THEN 1
        ELSE ROUND(c.percent_rank * 100, 0)
    END as percent_rank 
INTO #temp_top2_categories

FROM (
    SELECT c.customer_id,
        c.category_id,
        c.category_name,
        c.count_rentals,
        a.avg_rentals_category,
        c.percent_rank,
        m.max_count_rentals,
        p.total_rental_count_by_customer,
        ROW_NUMBER() OVER(PARTITION BY c.customer_id ORDER BY c.count_rentals DESC, c.lastest_rental_date DESC) as category_ranking
    FROM cte_rentals_by_customer c
    LEFT JOIN cte_avg_rental_by_category a ON a.category_id = c.category_id
    LEFT JOIN cte_max_rentals_by_category m ON m.category_id = c.category_id
    LEFT JOIN cte_pct_rental p ON p.customer_id = c.customer_id
) c
WHERE c.category_ranking <= 2


SELECT TOP 5 * FROM #temp_top2_categories 


-- CREATE TABLE: 1ST CATEGORY
DROP TABLE IF EXISTS #temp_1st_category;

SELECT c.customer_id,
    c.category_name,
    c.count_rentals,
    c.diff_rental_count_category,
    c.percent_rank
INTO #temp_1st_category
FROM #temp_top2_categories c
WHERE c.category_ranking = 1

SELECT TOP 5 * FROM #temp_1st_category

-- CREATE TABLE: 2ND CATEGORY
DROP TABLE IF EXISTS #temp_2nd_category;

SELECT c.customer_id,
    c.category_name,
    c.count_rentals,
    c.rental_share

INTO #temp_2nd_category
FROM #temp_top2_categories c
WHERE c.category_ranking = 2

SELECT TOP 5 * FROM #temp_2nd_category
```

```sql
-- CREATE TABLE: CATEGORY - FILM RECOMMENDATIONS
---- Create table: #temp_1st_category_film_recommedation
DROP TABLE IF EXISTS #temp_1st_category_film_recommendation;

WITH cte_top_film_rental AS (
    SELECT c.category_id,
        c.category_name,
        c.film_id,
        c.film_title,
        COUNT(c.rental_id) as count_rentals,
        ROW_NUMBER() OVER(PARTITION BY c.category_id, c.category_name ORDER BY COUNT(c.rental_id) DESC) as rental_ranking

    FROM #temp_category_section c 
    GROUP BY c.category_id, c.category_name, c.film_id, c.film_title
),
cte_1st_rcm AS (
    SELECT c.customer_id,
    c.category_name,
    f.film_title,
    f.rental_ranking,
    f.film_id

    FROM #temp_1st_category c
    LEFT JOIN cte_top_film_rental f 
    ON c.category_name = f.category_name
    WHERE f.rental_ranking <=3
)
SELECT *
INTO #temp_1st_category_film_recommendation
FROM cte_1st_rcm


SELECT TOP 5 * FROM #temp_1st_category_film_recommendation
ORDER BY customer_id ASC 


---- Create table: #temp_2nd_category_film_recommedation
DROP TABLE IF EXISTS #temp_2nd_category_film_recommendation;

WITH cte_top_film_rental AS (
    SELECT c.category_id,
        c.category_name,
        c.film_id,
        c.film_title,
        COUNT(c.rental_id) as count_rentals,
        ROW_NUMBER() OVER(PARTITION BY c.category_id, c.category_name ORDER BY COUNT(c.rental_id) DESC) as rental_ranking

    FROM #temp_category_section c 
    GROUP BY c.category_id, c.category_name, c.film_id, c.film_title
),
cte_2nd_rcm AS (
    SELECT c.customer_id,
    c.category_name,
    f.film_title,
    f.rental_ranking,
    f.film_id

    FROM #temp_2nd_category c
    LEFT JOIN cte_top_film_rental f 
    ON c.category_name = f.category_name
    WHERE f.rental_ranking <=3
)
SELECT *
INTO #temp_2nd_category_film_recommendation
FROM cte_2nd_rcm


SELECT TOP 5 * FROM #temp_2nd_category_film_recommendation
ORDER BY customer_id ASC 
```

```sql
-- CREATE TABLE: ACTOR & ACTOR RECOMMENDATION
---- Create table merge for actor figures
DROP TABLE IF EXISTS #temp_actor_section;

SELECT 
    r.rental_id,
    r.customer_id,
    a.actor_id,
    a.first_name,
    a.last_name,
    i.film_id,
    f.title,
    r.rental_date AS lastest_rental_date
INTO #temp_actor_section
FROM danny_mktcase.rental r 
LEFT JOIN danny_mktcase.inventory i ON r.inventory_id = i.inventory_id
LEFT JOIN danny_mktcase.film_actor fa ON fa.film_id = i.film_id
LEFT JOIN danny_mktcase.actor a ON a.actor_id = fa.actor_id
LEFT JOIN danny_mktcase.film f ON f.film_id = i.film_id
WHERE a.actor_id IS NOT NULL

SELECT TOP 5 * FROM #temp_actor_section
```

```sql
---- Create table: top actor rental count
DROP TABLE IF EXISTS #temp_top_actor;

WITH cte_top_actor_rental AS (

    SELECT a.customer_id,
            a.actor_id,
            COUNT(DISTINCT a.rental_id) as count_rentals,
            MAX(a.lastest_rental_date) as latest_rental_date,
            ROW_NUMBER() OVER(PARTITION BY a.customer_id ORDER BY COUNT(DISTINCT a.rental_id) DESC, MAX(a.lastest_rental_date) DESC) as actor_ranking

    FROM #temp_actor_section a
    GROUP BY a.customer_id, a.actor_id
)
SELECT DISTINCT r.customer_id,
    r.actor_id,
    a.first_name,
    a.last_name,
    r.count_rentals
INTO #temp_top_actor
FROM cte_top_actor_rental r
LEFT JOIN #temp_actor_section a ON r.customer_id = a.customer_id AND r.actor_id = a.actor_id
WHERE actor_ranking = 1

SELECT TOP 5 * FROM #temp_top_actor
ORDER BY customer_id ASC
```

```sql
---- Create table: TOP 3 film recommendations
DROP TABLE IF EXISTS #temp_actor_film_recommendation;
WITH cte_film_actor_count AS (
    SELECT a.film_id,
        a.actor_id,
        COUNT(DISTINCT a.rental_id) as count_rentals
    FROM #temp_actor_section a 
    GROUP BY a.film_id, a.actor_id

),
cte_filtered_category AS (
    SELECT *
    FROM cte_film_actor_count c
    WHERE NOT EXISTS 
    (
        SELECT 1 
        FROM #temp_1st_category_film_recommedation st
        WHERE c.film_id = st.film_id
    )
    AND NOT EXISTS
    (
        SELECT 1 
        FROM #temp_1st_category_film_recommedation nd
        WHERE c.film_id = nd.film_id
    )
),
cte_actor_film_rcm AS (
    SELECT DISTINCT a.customer_id,
        r.actor_id,
        r.count_rentals,
        r.film_id,
        s.title,
        r.film_rcm_ranking
    FROM (
        SELECT *,
            ROW_NUMBER() OVER(PARTITION BY c.actor_id ORDER BY c.count_rentals DESC) AS film_rcm_ranking
        FROM cte_filtered_category c
    ) r
    INNER JOIN #temp_top_actor a ON a.actor_id = r.actor_id
    INNER JOIN #temp_actor_section s ON s.film_id = r.film_id
    WHERE r.film_rcm_ranking <=3
)
SELECT * 
INTO #temp_actor_film_recommendation
FROM cte_actor_film_rcm 

SELECT TOP 10 * FROM #temp_actor_film_recommendation
ORDER BY customer_id ASC
```

```sql
DROP FUNCTION danny_mktcase.INITCAP;
GO

CREATE FUNCTION danny_mktcase.INITCAP(@input NVARCHAR(MAX))
RETURNS NVARCHAR(MAX)
AS
BEGIN
    DECLARE @output NVARCHAR(MAX);
    SELECT @output = 
        STRING_AGG(
            UPPER(LEFT(value, 1)) + LOWER(SUBSTRING(value, 2, LEN(value))),
            ' '
        )
    FROM STRING_SPLIT(@input, ' ');

    RETURN @output;
END
```

```sql
-- ====== FINAL TRANSFORMATIONS ======
DROP TABLE IF EXISTS  #temp_final_table;

WITH _1st_category AS(
    SELECT c.customer_id,
        c.category_name as cate1_name,
        c.count_rentals as cate1_rental_count,
        c.diff_rental_count_category as cate1_avg_comparison,
        c.percent_rank as cate1_percent_rank
    FROM #temp_1st_category c
),
_2nd_category AS(
    SELECT c.customer_id,
        c.category_name as cate2_name,
        c.count_rentals as cate2_rental_count,
        c.rental_share as cate2_rental_share
    FROM #temp_2nd_category c
),
_1st_rcm AS(
    SELECT c.customer_id,
        MAX(CASE WHEN c.rental_ranking = 1 THEN c.film_title END) cate1_reco1,
        MAX(CASE WHEN c.rental_ranking = 2 THEN c.film_title END) cate1_reco2,
        MAX(CASE WHEN c.rental_ranking = 3 THEN c.film_title END) cate1_reco3
    FROM #temp_1st_category_film_recommendation c
    GROUP BY c.customer_id
),
_2nd_rcm AS(
    SELECT c.customer_id as customer_id,
        MAX(CASE WHEN c.rental_ranking = 1 THEN c.film_title END) cate2_reco1,
        MAX(CASE WHEN c.rental_ranking = 2 THEN c.film_title END) cate2_reco2,
        MAX(CASE WHEN c.rental_ranking = 3 THEN c.film_title END) cate2_reco3
    FROM #temp_2nd_category_film_recommendation c
    GROUP BY c.customer_id
),
_top_actor AS(
    SELECT c.customer_id,
        CONCAT(c.first_name,' ',c.last_name) as actor_name,
        c.count_rentals as actor_rental_count
    FROM #temp_top_actor c
),
_actor_rcm AS(
    SELECT c.customer_id as customer_id,
        MAX(CASE WHEN c.film_rcm_ranking = 1 THEN c.title END) actor_reco1,
        MAX(CASE WHEN c.film_rcm_ranking = 2 THEN c.title END) actor_reco2,
        MAX(CASE WHEN c.film_rcm_ranking = 3 THEN c.title END) actor_reco3
    FROM #temp_actor_film_recommendation c
    GROUP BY c.customer_id
)
SELECT stc.customer_id,
        stc.cate1_name,
        danny_mktcase.INITCAP(str.cate1_reco1) cate1_reco1,
        danny_mktcase.INITCAP(str.cate1_reco2) cate1_reco2,
        danny_mktcase.INITCAP(str.cate1_reco3) cate1_reco3,
        ndc.cate2_name,
        danny_mktcase.INITCAP(ndr.cate2_reco1) cate2_reco1,
        danny_mktcase.INITCAP(ndr.cate2_reco2) cate2_reco2,
        danny_mktcase.INITCAP(ndr.cate2_reco3) cate2_reco3,
        danny_mktcase.INITCAP(ar.actor_reco1) actor_reco1,
        danny_mktcase.INITCAP(ar.actor_reco2) actor_reco2,
        danny_mktcase.INITCAP(ar.actor_reco3) actor_reco3,
        CONCAT('You''ve watched ', stc.cate1_rental_count, ' ',stc.cate1_name, ' films, that''s ',stc.cate1_avg_comparison, ' more than the DVD Rental Co average and puts you in the top ',stc.cate1_percent_rank,'% of ',stc.cate1_name,' gurus') 
        as insight_cate1,
        CONCAT('You''ve watched ', ndc.cate2_rental_count, ' ',ndc.cate2_name, ' films, making up ',ndc.cate2_rental_share, '% of your entire viewing history!') 
        as insight_cate2,
        CONCAT('You''ve watched ', ta.actor_rental_count, ' films featuring ', danny_mktcase.INITCAP(ta.actor_name), '! Here are some other films ',danny_mktcase.INITCAP(ta.actor_name), ' stars in that might interest you!') 
        as insight_actor
INTO #temp_final_table
FROM _1st_category stc
INNER JOIN _1st_rcm str ON stc.customer_id = str.customer_id
INNER JOIN _2nd_category ndc ON stc.customer_id = ndc.customer_id
INNER JOIN _2nd_rcm ndr ON stc.customer_id = ndr.customer_id
INNER JOIN _top_actor ta ON stc.customer_id = ta.customer_id
INNER JOIN _actor_rcm ar ON stc.customer_id = ar.customer_id

SELECT TOP 50 * FROM  #temp_final_table
ORDER BY customer_id ASC
```
